# 07 — Plotly and Dash

Ce notebook génère les figures Plotly, une prévisualisation et le script dashboard dans le dossier dashboard. Pour lancer une application Dash web permanente, exécuter le script sur une machine ou plateforme hébergée.

Source : jeu historique IBM DS0321EN. Exécuter toutes les cellules dans Google Colab. Les sorties sont produites à l’exécution, pas préremplies.

```python
%pip -q install plotly dash
import pandas as pd
import plotly.express as px
from pathlib import Path
from IPython.display import display
BASE="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/"
df=pd.read_csv(BASE+'dataset_part_2.csv')
assert {'LaunchSite','PayloadMass','Class','Orbit'}.issubset(df.columns)
OUT=Path('outputs');OUT.mkdir(exist_ok=True)
display(df.head())
```

```python
site=df.groupby('LaunchSite',as_index=False).agg(Launches=('Class','size'),SuccessRate=('Class','mean'))
fig=px.bar(site,x='LaunchSite',y='Launches',title='Falcon 9 Launches by Site')
fig.show(); fig.write_html(OUT/'dash_launches_by_site.html')
fig2=px.scatter(df,x='PayloadMass',y='Class',color='LaunchSite',hover_data=['Orbit'],
 title='Payload mass vs landing outcome (interactive)')
fig2.show();fig2.write_html(OUT/'dash_payload_scatter.html')
print('HTML charts saved. Screenshot figures for final report.')
```
